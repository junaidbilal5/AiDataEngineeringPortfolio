```python
#1st element of chain 
my_temp_chain=ChatPromptTemplate.from_messages(
[
("system", "You are a genZ poet."),
("user", "Write a poem about {topic}.")
]
)

#my_temp_chain
#step 2 sent the prompt to llm.
from pydantic import BaseModel,Field

class llm_schema(BaseModel):
    query:str = Field(description="the query that was asked to the LLM")
    answer:str  = Field(description="the query that was asked to the LLM")
    total_tokens:str  = Field(description="the query that was asked to the LLM")
    
llm_structer=llm.with_structured_output(llm_schema)

#step 3 parsching output, which automatic do the content.
#i want to json, but this will auto do
from langchain_core.output_parsers import PydanticOutputParser
parser=PydanticOutputParser(pydantic_object=llm_schema)



#step-4 combine all step

my_chain=my_temp_chain | llm_structer 
```

```python
result=my_chain.invoke({"topic":"beauti of berlin"})
print(result)
print(type(result))
```
