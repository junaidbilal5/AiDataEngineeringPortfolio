```python
from pydantic import BaseModel,Field
```

```python
class llm_schema(BaseModel):
    query:str = Field(description="the query that was asked to the LLM")
    answer:str  = Field(description="the query that was asked to the LLM")
    total_tokens:int  = Field(description="the query that was asked to the LLM")
    
```

```python
obj = llm_schema(
    query="what is the capital of France",
    answer="Paris",
    total_tokens=10
)

print(obj)
```

```python
obj = llm_schema(
    query="what is the capital of France",
    answer="Paris",
    total_tokens="total token is 10"
)

print(obj)
#error becuase token is int
```

```python
#now how we interigated with llm
llm_structured=llm.with_structured_output(llm_schema)
response=llm_structured.invoke("what is the capotal of france")
response

# we use llm structure when we need the final output
```
