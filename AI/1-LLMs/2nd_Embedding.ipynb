```python
from openai import OpenAI
client=OpenAI()


response=client.embeddings.create(
    input="what is ai",
    model="text-embedding-3-small" #bert model,  1536 because it is small
)

#print(response.data[0].embedding)
vector=response.data[0].embedding
len(vector)
```
